# Modular exponentiation via Montgomery product

```python
r: int = 2^64
```

```python
def montgomery_product(a: int, b: int, n: int, n_prime: int) -> int: 
    t: int = a * b
    m: int = (t * n_prime) % r
    u: int = (t + m * n) >> 64
    
    return u - n if u >= n else u
```

```python
def modular_multiplication(a: int, b: int, modulus: int) -> int:
    g, u, v = xgcd(r, modulus)
    assert g == 1
    n_prime: int = -v % r
    a_montgomery: int = (a << 64) % modulus
    b_montgomery: int = (b << 64) % modulus
    c_montgomery: int = montgomery_product(a_montgomery, b_montgomery, modulus, n_prime)
    c: int = montgomery_product(c_montgomery, 1, modulus, n_prime)

    return c
```

```python
print(modular_multiplication(12345, 67890, 187))
```

```python
def modular_exponentiation(M: int, e: int, modulus: int) -> int:
    g, u, v = xgcd(r, modulus)
    assert g == 1
    n_prime: int = -v % r

    M_montgomery: int = (M << 64) % modulus
    c_montgomery: int = (1 << 64) % modulus
    
    i: int = e.bit_length() - 1

    while i >= 0:
        c_montgomery = montgomery_product(c_montgomery, c_montgomery, modulus, n_prime)

        if ((e >> i) & 1) == 1:
            c_montgomery = montgomery_product(c_montgomery, M_montgomery, modulus, n_prime)
        
        i -= 1

    c = montgomery_product(c_montgomery, 1, modulus, n_prime)

    return c
```

```python
print(modular_exponentiation(7, 10, 13))
```
