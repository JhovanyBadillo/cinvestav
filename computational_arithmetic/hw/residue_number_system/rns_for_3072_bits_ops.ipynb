# Residue Number System

Find a basis of $k$ 64-bit pairwise relatively prime moduli such that $M = \Pi_{i=1}^{k} m_i > 2^{3072} - 1$.
With these moduli it would be possible to represent all unsigned integers of $3072$ bits. For convenience, we consider only primes.

```python
top = 2^3072 - 1
```
