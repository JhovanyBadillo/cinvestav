# Diffie-Hellman secret sharing

## Some functions

```python
def _compute_orders(p: int) -> dict[int, int]:
    orders: dict[int, int] = {}

    for a in range(1, p):
        for i in range(1, p):
            q = (a^i) % p
            if (q == 1) and not a in orders:
                orders[a] = i

    return orders
```

```python
def _compute_generators(p: int) -> list[int]:
    # a generator has order p - 1
    orders: dict[int, int] = _compute_orders(p)
    return [g for g in orders if orders[g] == (p - 1)]
```

```python
import random

def get_random_generator(p: int) -> int:
    return random.choice(_compute_generators(p))
```
